```python

import pandas as pd

# Replace this path with the one you copied
file_path = "/content/googleplaystore.csv"

# Load the dataset
df = pd.read_csv(file_path)

# Display the first 5 rows
df.head()
```

```python

# Check the dataset size
print("DATASET SHAPE:")
print(df.shape)

# Display all column names
print("\nCOLUMN NAMES:")
print(df.columns.tolist())

# Check missing values
print("\nMISSING VALUES:")
print(df.isnull().sum())
```

```python
print("Duplicate app names:", df["App"].duplicated().sum())
```

```python

# Count rows that are identical across all columns
exact_duplicates = df.duplicated().sum()

print("Exact duplicate rows:", exact_duplicates)
```

```python

# Remove exact duplicate rows
df = df.drop_duplicates().copy()

# Check the updated dataset
print("Rows remaining:", len(df))
print("Exact duplicates remaining:", df.duplicated().sum())
```

```python

# Check missing values after removing duplicates
print("MISSING VALUES:")
print(df.isnull().sum())
```

```python

# Check the data types of each column
print("DATA TYPES:")
print(df.dtypes)
```

```python

# Inspect sample values before conversion
for column in ["Reviews", "Installs", "Price"]:
    print(f"\n{column}:")
    print(df[column].value_counts().head(10))
```

```python

import pandas as pd

# Remove malformed rows
df = df[df["Category"] != "1.9"].copy()

# Convert reviews to numbers
df["Reviews"] = pd.to_numeric(
    df["Reviews"], errors="coerce"
)

# Clean installs
df["Installs"] = pd.to_numeric(
    df["Installs"].astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False),
    errors="coerce"
)

# Clean prices
df["Price"] = pd.to_numeric(
    df["Price"].astype(str)
    .str.replace("$", "", regex=False),
    errors="coerce"
)

# Convert update dates
df["Last Updated"] = pd.to_datetime(
    df["Last Updated"], errors="coerce"
)

# Fill missing ratings with the median
df["Rating"] = df["Rating"].fillna(
    df["Rating"].median()
)

# Check results
print("CLEANING COMPLETE!")
print("Rows:", len(df))
print("\nData Types:")
print(df[["Reviews", "Installs", "Price"]].dtypes)
print("\nMissing Ratings:", df["Rating"].isna().sum())
```

```python

# GOOGLE PLAY STORE - FINAL ANALYSIS

print("TOP 10 APP CATEGORIES")
print(df["Category"].value_counts().head(10))

print("\nAVERAGE APP RATING")
print(round(df["Rating"].mean(), 2))

print("\nTOP 10 MOST INSTALLED APPS")
print(
    df[["App", "Installs"]]
    .sort_values("Installs", ascending=False)
    .head(10)
)

print("\nFREE VS PAID APPS")
print(df["Type"].value_counts())
```
